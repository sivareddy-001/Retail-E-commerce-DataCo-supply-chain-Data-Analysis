```python
df['Order Item Total'] = df['Order Item Quantity'].fillna(df['Order Item Total'].median())
```

```python
df.columns
```

```python
df.rename(columns=
 {
     'Late_delivery_risk': 'Late Delivery Risk',
     'order date ': 'Order Date',
     'shipping date': 'Shipping Date'
 }, inplace=True)
```

```python
df.columns
```

```python
category_columns = [
    'Type',
    'Delivery Status',
    'Category Name',
    'Customer City',
    'Customer Segment',
    'Department Name',
    'Market',
    'Order Country',
    'Order Region',
    'Order Status',
    'Product Name',
    'Shipping Mode',
]
df[category_columns] = df[category_columns].fillna(df[category_columns].mode().iloc[0])
```

```python
numerical_cols = [
    'Days for shipping (real)',
    'Days for shipment (scheduled)',
    'Late Delivery Risk',
    'Order Profit Per Order',
    'Inventory Stock (Units)'
]
df[numerical_cols] = df[numerical_cols].fillna(df[numerical_cols].mean())
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
Total_sales = df['Sales'].sum()
print("Total_sales:", Total_sales)
```

```python
Total_Orders = df['Order ID'].count()
print("Total_Orders:", Total_Orders)
```

```python
Avg_delivery_days = df['Days for shipping (real)'].mean()
print("Avg_delivery_days:", round(Avg_delivery_days,2))
```

```python
On_time_orders = (df["Delivery Status"] == "Shipping on time").sum()
Total_Orders = df['Order ID'].count()
On_time_percentage = (On_time_orders/Total_Orders)*100
print("On_time_percentage :", round(On_time_percentage,2), "%")
```

```python
df.to_excel("cleaned_data.xlsx", index=False)
```
